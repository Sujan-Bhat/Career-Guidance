# CAREERMIND Pilot Study

Planned analyses (paper Sec. VIII):

1. **FES predictive validity** - correlation of FES (and each sub-metric) against held-out academic outcomes; per-student vs population weight vectors.
2. **Recommender benchmark** - 3-stage cascade vs kg_only / cf_only / fm_only / popularity baselines: Precision@10, Recall@10, nDCG@10.
3. **DQN convergence** - training reward curves, epsilon schedule, policy stability under the Eq. 3 reward.
4. **Sensitivity analysis** - sweep design thresholds listed in `evaluation/sensitivity.py`.

Cells are implemented in Phase 9.
