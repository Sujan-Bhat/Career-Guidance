# CAREERMIND Pilot Study

The planned analyses (paper Sec. VIII) are implemented as a runnable CLI in
`evaluation/run_pilot.py` — this notebook is kept as the study plan/pointer.

```bash
make eval      # PYTHONPATH=ml:evaluation python evaluation/run_pilot.py
```

1. **FES predictive validity** — correlation of FES (and each sub-metric)
   against held-out academic outcomes; per-student vs population weights.
2. **Recommender benchmark** — leave-one-out P@10 / R@10 / nDCG@10 / Hit@10
   for the 3-stage cascade vs kg_only / cf_only / fm_only / popularity,
   plus the stage-1 gate pass-rate diagnostic.
3. **DQN convergence** — reward curve read from
   `ml/artifacts/dqn_training.json` (Eq. 3 training run).
4. **Sensitivity analysis** — sweeps over the stage-1 eligibility gate,
   EngagementSignal thresholds, the ≥8-session calibration gate, and Eq. 3
   reward weights (α/β/γ/δ).

Results are written to `evaluation/results/pilot_results.json`.
Supporting modules: `benchmark.py`, `sensitivity.py`, `baselines.py`,
`metrics.py`; tests under `evaluation/tests/` (`make test` runs them).
